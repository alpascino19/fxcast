```python
import pandas as pd

f = pd.read_csv("../data/features.csv", index_col="date", parse_dates=True)
f = f.dropna(subset=["target_big"])

for col in ["politic_ratio_3d", "total_ratio_3d", "EUR_vol_5", "EUR_maxabs_5"]:
    groups = pd.qcut(f[col], 5, duplicates="drop")
    print(f"\n{col}")
    print((f.groupby(groups, observed=True)["target_big"].mean() * 100).round(1))
```

```python
recent = f.loc["2019":]
calm = recent[recent["EUR_vol_5"] < recent["EUR_vol_5"].median()]

print(f"Baza 2019+: {recent['target_big'].mean() * 100:.1f}% din {len(recent)} zile")
print(f"Baza in zile calme: {calm['target_big'].mean() * 100:.1f}% din {len(calm)} zile\n")

for name, data in [("2019+", recent), ("zile calme", calm)]:
    for col in ["politic_ratio_3d", "total_ratio_3d"]:
        for q in [0.80, 0.90, 0.95, 0.98]:
            thr = data[col].quantile(q)
            sub = data[data[col] > thr]
            print(f"{name:10} {col:17} top {(1 - q) * 100:2.0f}% (>{thr:4.1f}): "
                  f"{sub['target_big'].mean() * 100:5.1f}% din {len(sub):3d} zile")
        print()
```
