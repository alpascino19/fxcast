```python
import pandas as pd

news = pd.read_csv("../data/news.csv")
sources = pd.read_csv("../data/sources.csv")

counts = news["source"].value_counts()
missing = [s for s in counts.index if s not in set(sources["source"]) and counts[s] >= 3]
print(f"Surse noi neclasificate: {len(missing)}")
for s in missing:
    print(f"{s}: {counts[s]} articole")
```

```python
counts = news["source"].value_counts()
top = counts[counts >= 3]

skeleton = pd.DataFrame({
    "source": top.index,
    "articles": top.values,
    "tier": "",
    "type": "",
    "owner_group": "",
    "trust_dnr_2026": "",
    "jti": "",
    "notes": "",
})

skeleton.to_csv("../data/sources.csv", index=False, encoding="utf-8")
print(f"Surse in fisier: {len(skeleton)}")
print(f"Acopera {top.sum() / counts.sum() * 100:.1f}% din articole")
```
